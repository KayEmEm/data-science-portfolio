# Run one portfolio project in Colab

This notebook runs any of the 13 projects with Google Drive as the only storage. The code comes
from GitHub on each run; the data, checkpoints and results stay in Drive.

**Before the first run**

1. In Drive, create `MyDrive/portfolio/raw/<project folder>/` and put the raw files of the project
   there, as listed in `colab/README.md` of the repository. Projects that download their own data
   need nothing there.
2. For p06, which trains on a GPU, select Runtime, Change runtime type, T4 GPU.
3. Run all cells. Results appear in `MyDrive/portfolio/out/<project folder>/`, copied every
   10 minutes during the run and once at the end.

Colab asks for access to Drive when the mount cell runs: that access covers the whole Drive.

```python
#@title Settings
PROJECT = 'p01'  #@param ['p01', 'p02', 'p03', 'p04', 'p05', 'p06', 'p07', 'p08', 'p09', 'p10', 'p11', 'p12', 'p13']
PROFILE = 'free'  #@param ['free', 'pro']
REF = 'colab-v3'  #@param {type:'string'}
RESUME = False  #@param {type:'boolean'}
TARGET = ''  #@param {type:'string'}
SAMPLE = ''  #@param {type:'string'}
DRIVE_ROOT = '/content/drive/MyDrive/portfolio'  #@param {type:'string'}
# REF: the reviewed tag that this notebook runs (see colab/README.md); 'main' runs the latest code. TARGET and SAMPLE: empty keeps the profile values.
# RESUME: copy the outputs saved in Drive back first, so that steps that resume can continue.
```

```python
#@title Machine
import shutil
import subprocess

# nvidia-smi exists only on a GPU runtime.
gpu = subprocess.run(['nvidia-smi', '-L'], capture_output=True, text=True).stdout.strip() if shutil.which('nvidia-smi') else ''
print(gpu or 'No GPU in this runtime: CPU only')
print(subprocess.run(['python', '--version'], capture_output=True, text=True).stdout.strip())
```

```python
#@title Google Drive
import os

from google.colab import drive

drive.mount('/content/drive')
os.makedirs(f'{DRIVE_ROOT}/raw', exist_ok=True)
print(sorted(os.listdir(f'{DRIVE_ROOT}/raw')))
```
